**🎯 Project Requirements**

Create a food delivery system with different payment methods:

Cash on Delivery,
Credit Card,
Mobile Wallet

All payment methods should have a common method:

```python
from abc import ABC, abstractmethod
```

**Step 2: Create Abstract Class**

Create a class called Payment.

```python
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass
```

**Step 3: Create Cash Payment**

```python
class CashPayment(Payment):

    def pay(self, amount):
        print(f"Cash payment of Rs. {amount} received.")
```

**Step 4: Create Card Payment**

```python
class CardPayment(Payment):

    def pay(self, amount):
        print(f"Credit card payment of Rs. {amount} processed.")
```

**Step 5: Create Wallet Payment**

```python
class WalletPayment(Payment):

    def pay(self, amount):
        print(f"Mobile wallet payment of Rs. {amount} processed.")
```

**Step 8: Complete Solution**

Here is the complete project:

```python
from abc import ABC, abstractmethod


# Abstract Class
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# Cash Payment
class CashPayment(Payment):

    def pay(self, amount):
        print(f"Cash payment of Rs. {amount} received.")


# Credit Card Payment
class CardPayment(Payment):

    def pay(self, amount):
        print(f"Credit card payment of Rs. {amount} processed.")


# Mobile Wallet Payment
class WalletPayment(Payment):

    def pay(self, amount):
        print(f"Mobile wallet payment of Rs. {amount} processed.")


# Create objects
cash = CashPayment()
card = CardPayment()
wallet = WalletPayment()


# Make payments
cash.pay(1500)
card.pay(2500)
wallet.pay(1000)
```
